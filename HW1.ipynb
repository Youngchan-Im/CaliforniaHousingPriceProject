```python
housing = pd.read_csv("housing.csv")

def convert(x):
    x = x.replace("$", "")
    x = x.replace(",", "")
    return float(x)

housing["median_house_value"] = housing["median_house_value"].apply(convert)

housing.info()
```

```python
housing_clean = housing.dropna()

housing_clean.info()
```

```python
housing_clean.describe()

# Negative --> longitude --> not problem (it can be negative..?)
# Cap value --> median_income & median_house_value
```

```python
select = housing_clean[['median_house_value', 'median_income', 'housing_median_age']]
sns.pairplot(select)

select.describe()
```

```python
a = housing_clean["median_house_value"]

names = ["mean", "median", "std", "min", "max"]
values = [a.mean(), a.median(), a.std(), a.min(), a.max()]

plt.bar(names, values)
plt.show()
```
